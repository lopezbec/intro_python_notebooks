## 2. From a Python list to a Pandas Series

You already know how to store data in a **list**. A Pandas **Series** is similar: it stores a sequence of values, but it also displays an **index** next to each value.

In our first examples, the index is the familiar `0, 1, 2, 3` sequence, so we can use ordinary indexing and slicing just as we did with lists.

````python
# A familiar Python list
highs = [72, 75, 78, 81]
print("Python list:", highs)

# The same numbers as a Pandas Series
high_series = pd.Series(highs)
print("\nPandas Series:")
print(high_series)
````

### 2.1 Selecting values with familiar square brackets

For our Series with the default numbered index, the syntax looks like a Python list:

| Expression | Meaning |
|---|---|
| `high_series[0]` | First temperature |
| `high_series[2]` | Third temperature |
| `high_series[1:3]` | Second and third temperatures |

**Hint:** The slice `[1:3]` includes index `1`, but **not** `3`. There is no new row-selection method to learn for these examples.

Later, you may see other indexing tools when working with more advanced Pandas tables. We do not need them yet.

````python
print("First high:", high_series[0])
print("\nSecond and third highs:")
print(high_series[1:3])
````

### 2.2 A Series can also use text labels

A Python **dictionary** connects a *key* to a *value*. A Pandas Series can use those dictionary keys as **labels** instead of numbers.

For example, the label `"Tuesday"` lets us look up Tuesday's temperature. This is similar to using `my_dictionary["Tuesday"]`.

**Hint:** If a Series has day names as its labels, use the day name to select a value. The simple numbered examples from Section 2.1 have a different type of index.

````python
# Dictionary: day -> high temperature
highs_by_day = {
    "Monday": 72,
    "Tuesday": 75,
    "Wednesday": 78,
    "Thursday": 81
}

daily_highs = pd.Series(highs_by_day)
print(daily_highs)
print("\nTuesday:", daily_highs["Tuesday"])
````

## 3. From dictionaries to a DataFrame

A **DataFrame** is a table. We can create one using a dictionary in which:

- Each **key** becomes a column name.
- Each **list** contains the values in that column.
- Each list should have the **same number of items**.

Let's make a small weather table using the same numbers.

````python
weather_dict = {
    "Day": ["Monday", "Tuesday", "Wednesday", "Thursday"],
    "High": [72, 75, 78, 81],
    "Low": [51, 54, 56, 58]
}

small_weather = pd.DataFrame(weather_dict)
small_weather
````

### 3.1 Reading a DataFrame

The **column names** appear across the top. The **index** appears on the left and identifies each row; by default it starts at `0`.

Try these common commands:

| Command | What it tells you |
|---|---|
| `.head()` | First five rows |
| `.head(2)` | First two rows |
| `.shape` | Number of rows and columns `(rows, columns)` |
| `.columns` | Names of the columns |
| `.index` | Labels used for rows |

**Hint:** `.head()` has parentheses because it is a method; `.shape` does **not**.

````python
print("Table size (rows, columns):", small_weather.shape)
print("Column names:", list(small_weather.columns))
print("Row labels:", small_weather.index)

small_weather.head(2)
````

### 3.2 One column versus several columns

Use the column **name** between square brackets.

- `small_weather["High"]` selects **one** column and produces a **Series**.
- `small_weather[["Day", "High"]]` selects **two** columns and produces a **DataFrame**.

**Hint:** Notice the **double square brackets** when selecting more than one column.

````python
print("One column:")
print(small_weather["High"])

print("\nTwo columns:")
small_weather[["Day", "High"]]
````

### 3.3 Select rows with ordinary slices

For a **DataFrame**, a slice such as `small_weather[1:3]` selects **rows**. It works like Python list slicing: the ending position is not included.

| Expression | Rows shown |
|---|---|
| `small_weather[0:1]` | First row |
| `small_weather[0:2]` | First two rows |
| `small_weather[1:3]` | Second and third rows |

**Remember:** A quoted name such as `small_weather["High"]` selects a **column**; a numbered slice such as `small_weather[1:3]` selects **rows**. For now, our example table uses the default row numbering.

````python
print("First row:")
print(small_weather[0:1])

print("\nSecond and third rows:")
small_weather[1:3]
````

### 3.4 Combine row slicing and column selection

- `[1:4]` keeps the second, third, and fourth rows.
- `[["Day", "High"]]` keeps just the two named columns.

**Hint:** Read expressions from left to right: *choose rows* → *choose columns*.

## 4. Create, change, and filter information

A DataFrame can hold *new information that we calculate* from existing columns.

For instance, the **daily temperature range** is the high temperature minus the low temperature:

`Range = High - Low`

````python
# Create a new column using two existing columns.
small_weather["Range"] = small_weather["High"] - small_weather["Low"]
small_weather
````

### 4.1 Change a column

We can update an **entire column** with a calculation. For example, suppose the high temperature on each day were **1°F warmer**.

The next cell makes a **copy** of our table, increases every value in the `High` column by one, and recalculates the `Range` column.

**Hint:** This is simpler than editing individual cells, and it shows a useful Pandas idea: calculations can work on a whole column.

````python
updated_weather = small_weather.copy()

# Increase each high temperature by one degree.
updated_weather["High"] = updated_weather["High"] + 1

# Update the range after changing the highs.
updated_weather["Range"] = updated_weather["High"] - updated_weather["Low"]

updated_weather
````

### 4.2 Filter: keep rows that meet a condition

A **filter** lets us focus on rows we care about.

For example, `small_weather["High"] > 75` asks a question for every row and gives `True` or `False`.

We can use those True/False values to keep only matching rows.

````python
print("Does each day's high exceed 75°F?")
print(small_weather["High"] > 75)

print("\nOnly the matching days:")
small_weather[small_weather["High"] > 75]
````

## 6. Understand data types and missing values

When we read a file, some columns may be stored as text even if they represent dates or numbers.

Pandas gives us two useful tools:

- **`.dtypes`** displays the type of values in each column.
- **`pd.to_datetime()`** changes date-like text into dates.

For temperatures, `pd.to_numeric(..., errors="coerce")` converts values into numbers. If it finds a blank or an invalid value, it stores a **missing value** instead of stopping.

````python
print("Before preparing the columns:")
print(weather_2021_simple[["DATE", "TMAX", "TMIN"]].dtypes)

# Change the existing columns into useful data types.
weather_2021_simple["DATE"] = pd.to_datetime(weather_2021_simple["DATE"])
weather_2021_simple["TMAX"] = pd.to_numeric(weather_2021_simple["TMAX"], errors="coerce")
weather_2021_simple["TMIN"] = pd.to_numeric(weather_2021_simple["TMIN"], errors="coerce")

print("\nAfter preparing the columns:")
print(weather_2021_simple[["DATE", "TMAX", "TMIN"]].dtypes)
````

### 6.1 Missing values

Real datasets can have blank measurements. Pandas often displays missing numbers as **`NaN`**, which is a marker for a missing value (not the number zero).

- `.isna()` identifies missing entries.
- `.isna().sum()` counts them by column.
- `.dropna()` makes a new table that **excludes rows with missing values** in chosen columns.

**Important:** A missing high temperature doesn't mean the temperature was `0°F`. We shouldn't replace it with zero.

````python
print("Missing measurements:")
print(weather_2021_simple[["TMAX", "TMIN"]].isna().sum())

# Keep rows that contain both a high and a low temperature.
weather_ready = weather_2021_simple.dropna(subset=["TMAX", "TMIN"]).copy()

print("\nOriginal number of rows:", len(weather_2021_simple))
print("Rows after removing missing temperatures:", len(weather_ready))
weather_ready[["DATE", "TMAX", "TMIN"]].head()
````
