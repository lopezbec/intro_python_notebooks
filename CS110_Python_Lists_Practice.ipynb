## Setup: Automatic Checkers

Run the code cell below once. It creates helper functions that the later exercises use to check your work.

````python
def check_value(variable_name, expected):
    """Check whether a variable exists and has the expected value."""
    if variable_name not in globals():
        print(f"❌ {variable_name} has not been created yet.")
        return False

    actual = globals()[variable_name]
    if actual == expected:
        print(f"✅ {variable_name} is correct!")
        return True

    print(f"❌ {variable_name} is not correct yet.")
    print("   Current value:", actual)
    print("   Re-read the instructions, revise your code, and try again.")
    return False


def check_list(variable_name, expected):
    """Check that a variable is a list and has the expected contents."""
    if variable_name not in globals():
        print(f"❌ {variable_name} has not been created yet.")
        return False

    actual = globals()[variable_name]
    if not isinstance(actual, list):
        print(f"❌ {variable_name} exists, but it is not a list.")
        print("   Current value:", actual)
        return False

    return check_value(variable_name, expected)


print("Checker functions are ready.")
````

# 1. Defining and Accessing a List

A **list** lets us store multiple related values in one variable.

Lists use square brackets `[]`, and the items are separated by commas.

```python
students = ['EJ', 'Aaron', 'Daniel', 'Veronica']
```

Each item has a position called an **index**. Python begins counting at **0**.

| Item | `EJ` | `Aaron` | `Daniel` | `Veronica` |
|---|---:|---:|---:|---:|
| Index | `0` | `1` | `2` | `3` |

To access one item, place its index inside square brackets:

```python
print(students[0])
```

This prints the first item in the list.

````python
# Example: create and access a list
students = ['EJ', 'Aaron', 'Daniel', 'Veronica', 'Alivia', 'Emma']

print(students)
print('First student:', students[0])
print('Third student:', students[2])
````

# 3. Selecting Multiple Items with Slicing

A **slice** selects several items from a list at once.

The basic syntax is:

```python
my_list[start_index:end_index]
```

The `start_index` is included, but the `end_index` is **not included**.

For example:

```python
numbers = [10, 20, 30, 40, 50]
numbers[1:4]
```

produces:

```python
[20, 30, 40]
```

Useful shortcuts:

- `my_list[:3]` → from the beginning through index `2`
- `my_list[3:]` → from index `3` through the end
- `my_list[-2:]` → the last two items

````python
# Example: slicing
students = ['EJ', 'Aaron', 'Dan', 'Veronica', 'Alivia', 'Emma']

print('First two:', students[:2])
print('Middle two:', students[2:4])
print('Last two: ', students[-2:])
````

# 4. Modifying Lists with Methods

Python lists have built-in **methods** that can add, remove, or rearrange items.

### Common methods

| Method | Purpose | Example |
|---|---|---|
| `.append(item)` | Add an item to the end | `names.append('Aidan')` |
| `.insert(index, item)` | Add an item at a specific index | `names.insert(1, 'Mia')` |
| `.remove(item)` | Remove an item by value | `names.remove('EJ')` |
| `.sort()` | Sort the list in place | `names.sort()` |

These methods **change the existing list**.

````python
# Example: list methods
students = ['Veronica', 'Aaron', 'EJ']
print('Start:  ', students)

students.append('Aidan')
print('Append: ', students)

students.insert(1, 'Emma')
print('Insert: ', students)

students.remove('EJ')
print('Remove: ', students)

students.sort()
print('Sorted: ', students)
````

# 5. Combining Lists with `+`

Two lists can be joined with the `+` operator.

This is similar to combining strings, but the result is a new list containing the items from both lists.

```python
left_side = ['Jack', 'Casey', 'Ella']
right_side = ['Jayson', 'Milan', 'Rawle']

front_row = left_side + right_side
```

The order matters: the items from the first list come first, followed by the items from the second list.

````python
# Example: combine two lists
left_side = ['Jack', 'Casey', 'Ella']
right_side = ['Jayson', 'Milan', 'Rawle']

front_row = left_side + right_side
print(front_row)
````
