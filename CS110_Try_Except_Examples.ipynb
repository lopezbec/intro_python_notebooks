# CS110 — When Programs Go Wrong: `try` / `except`

**Goal:** Learn how to handle common Python runtime errors without stopping the entire program.

Run the code cells **one at a time**. When a cell asks for input, run it again with the suggested values and compare the results.

**Remember:**
- `try`: code that might cause an error.
- `except`: what to do if a **specific** error happens.
- After an exception is handled, Python continues **after** the `try/except` block.

## 1. Prevent an error with `if`

Sometimes we can check a value **before** performing a risky operation. This is the approach shown in the slides before introducing `try/except`.

**Try this:** Change `count` from `0` to `3` and run again.

```python
total = 12
count = 0  # Change to 3 and run again

if count != 0:
    print("Result:", total / count)
else:
    print("Cannot divide by zero.")
```

## 2. Your first `try/except`: `ValueError`

`int()` raises a `ValueError` when text cannot be converted to an integer.

**Try these inputs:** `7`, then `hello`. Notice whether the `except` block runs, and whether the last line still prints.

```python
try:
    number = int(input("Enter a whole number: "))
    print("Double:", number * 2)
except ValueError:
    print("Please enter a whole number.")

print("This example has finished.")
```

## 3. Catch a `ZeroDivisionError`

Division by zero causes a runtime exception. `except ZeroDivisionError` lets us show a useful message instead.

**Try this:** Change `divisor` to `4` and run again.

```python
divisor = 0  # Change to 4

try:
    result = 12 / divisor
    print("Result:", result)
except ZeroDivisionError:
    print("Cannot divide by zero.")

print("The program keeps running.")
```

## 4. One `try` block, two possible errors

A single `try` can have **more than one `except`**. Python uses the one that matches the exception.

**Try these inputs:** `3` (works), `abc` (`ValueError`), and `0` (`ZeroDivisionError`).

```python
try:
    number = int(input("Divisor: "))
    print("Result:", 12 / number)
except ValueError:
    print("Please enter an integer.")
except ZeroDivisionError:
    print("Cannot divide by zero.")
```

## 5. Another example: `KeyError` with a dictionary

A `KeyError` occurs when we ask for a dictionary key that does not exist.

**Try this:** Replace `"B"` with `"A"` to see the successful case.

```python
scores = {"A": 9}

try:
    print("Score:", scores["B"])
except KeyError:
    print("That key is not in the dictionary.")
```

## 6. Handling an error is **not** the same as retrying

An `except` block handles the error **once**. It does not ask for another input automatically.

**Try these inputs:** `19`, then `apple`. Observe that the program finishes in both cases.

```python
try:
    age = int(input("Age: "))
    print("Age entered:", age)
except ValueError:
    print("Not a valid number.")

print("Program finished.")
```

## 7. Use a `while` loop to try again

To ask the user again, combine a loop with `try/except`.

- `continue` starts the next loop iteration after invalid input.
- `break` exits the loop after valid input.

**Try this sequence:** `hello`, then `19`.

```python
while True:
    try:
        age = int(input("Age: "))
    except ValueError:
        print("Enter a whole number.")
        continue  # Ask again

    break  # The conversion worked

print("Age saved:", age)
```

## 8. An integer can still be an invalid answer

`try/except` answers **“Can we convert the input?”** An `if` statement answers **“Does the value make sense?”** We often need both.

**Try this sequence:** `apple`, then `-2`, then `20`.

```python
while True:
    try:
        age = int(input("Age: "))
    except ValueError:
        print("Enter a whole number.")
        continue

    if age < 0:
        print("Age cannot be negative.")
        continue

    break  # Valid integer and a nonnegative age

print("Age saved:", age)
```

## What to remember

1. Catch the **specific exception** you expect, such as `ValueError` or `ZeroDivisionError`. Avoid using a bare `except:` to hide other bugs.
2. `except` handles an error, but **does not retry** by itself. Use a loop when another attempt is needed.
3. Use `if` statements to check whether a converted value is **reasonable**.
4. Test your programs with **normal, edge-case, and invalid** inputs.
