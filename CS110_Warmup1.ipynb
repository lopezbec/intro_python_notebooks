## Optional: check all 12 problems

This cell runs all 12 individual testers and summarizes the results. **You do not need it to work problem by problem.** Run it only after defining all 12 exercise functions, or use it to see which ones still need work.


```python

# Each entry contains the student's function name and its test function.
tests = [
    ("sleep_in", sleep_in_test),
    ("monkey_trouble", monkey_trouble_test),
    ("sum_double", sum_double_test),
    ("diff21", diff21_test),
    ("parrot_trouble", parrot_trouble_test),
    ("makes10", makes10_test),
    ("near_hundred", near_hundred_test),
    ("pos_neg", pos_neg_test),
    ("not_string", not_string_test),
    ("missing_char", missing_char_test),
    ("front_back", front_back_test),
    ("front3", front3_test)
]

results = []

# Run all tests.
for name, test_function in tests:

    student_function = globals().get(name)

    if callable(student_function):
        passed, total = test_function(student_function)
        results.append((name, passed, total))
    else:
        print(f"\n{name}: Function not defined.")
        results.append((name, 0, 0))


# ============================================================
# Overall Report
# ============================================================

print("\n" + "=" * 55)
print("CS110 WARMUP-1 — OVERALL REPORT")
print("=" * 55)

for name, passed, total in results:

    if total == 0:
        print(f"{name:<18} NOT DEFINED")
    else:
        status = "PASS" if passed == total else "NEEDS WORK"
        print(f"{name:<18} {passed:>2}/{total:<2} {status}")

print("-" * 55)

print(f"TOTAL: {sum(p for _, p, t in results)}/{sum(t for _, p, t in results)} tests passed")

print("Note: CodingBat's undisclosed tests are not included.")
```
