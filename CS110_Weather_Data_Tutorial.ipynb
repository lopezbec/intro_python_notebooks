## 2. Open the File with Python Built-In Functions

*   The `open()` function takes the path as an argument and returns a file object
*   From the file object, you can call the `read()` method, which returns the entire file as a string



```python
my_file = open(file_path)
one_big_string = my_file.read()
print(one_big_string)
my_file.close()
```

```python
# Alternative Syntax
with open(file_path) as my_file:
  one_big_string = my_file.read()
  print(one_big_string)
```

## 3. Read the File _Line-by-Line_ with Python Built-In Functions

*   We can iterate over each line in a file, similar to how we can iterate over each item in a list!
*   This can be done with the keywords `for` and `in`



```python
with open(file_path) as my_file:
  for line in my_file:
    print('==>', line)
```

## 4. Read the File into a List with Python Built-In Functions

* In the previous example, each line is a **_string_**. However, to access individual pieces of data, it is useful to extract/split up the string.
* The Python `split()` method for strings can help us do this! It splits the string into a list based on the specified separator.



**Important:** The `NAME` field contains a comma inside quotation marks. Basic `split(",")` will treat that comma as a separator, so the example below **does not correctly parse every column**. Use `csv.reader()` or `csv.DictReader()` (Sections 5–6) for accurate CSV processing.

```python
with open(file_path) as my_file:
  for line in my_file:
    line_lst = line.split(",")  # Turn the string into a list manually
    print(line_lst[0], '--', line_lst[1], '--', line_lst[2], '--', line_lst[3] )
```

## 5. Read the File into a List with Python's CSV Library

* Reading and parsing files is common, so Python provides a standard library to do this more easily
* The library is called `csv`
* The library function `csv.reader()` is used to streamline reading from an open CSV file
* The function `next()` allows us to skip a row of the file; Useful for if we have a header!



```python
import csv

with open(file_path) as my_file:
  my_csv = csv.reader(my_file)
  for line_lst in my_csv:
    print(line_lst[0], '--', line_lst[1], '--', line_lst[2], '--', line_lst[3] )
```

## 6. Read the File into a Dictionary with Python's CSV Library

* The `csv` library function `csv.DictReader()` produces a dictionary instead of a list! This enables you to access data values by their header name, rather than numbered position

```python
import csv

with open(file_path) as my_file:
  my_csv = csv.DictReader(my_file)
  for line_dict in my_csv:
    print(line_dict['STATION'], '--', line_dict['NAME'],
          '--', line_dict['DATE'], '--', line_dict['TMAX'])
```
