## Example 3 — More Than One Turtle

You can create more than one Turtle object. Here, `tess` and `alex` are **different objects**, so each one can receive its own instructions.

Watch which turtle draws each shape:
- Tess draws a triangle.
- Alex draws a square.

### Try changing it
- Give Alex a color and pen size.
- Change Tess's triangle size.
- Move one turtle before it begins drawing.
- **Challenge:** Can Tess draw the square while Alex draws the triangle instead?

````python
# CHANGED: Use ColabTurtlePlus instead of Python's standard turtle library.
import ColabTurtlePlus.Turtle as turtle

# ADDED: Clear any previous Turtle drawing from the Colab output.
turtle.clearscreen()


wn = turtle.Screen()             # Set up the window and its attributes
wn.bgcolor("lightgreen")


tess = turtle.Turtle()           # create tess and set some attributes
tess.color("hotpink")
tess.pensize(5)

alex = turtle.Turtle()           # create alex


tess.forward(80)                 # Let tess draw an equilateral triangle
tess.left(120)
tess.forward(80)
tess.left(120)
tess.forward(80)
tess.left(120)                   # complete the triangle


tess.right(180)                  # turn tess around
tess.forward(80)                 # move her away from the origin


alex.forward(50)                 # make alex draw a square
alex.left(90)
alex.forward(50)
alex.left(90)
alex.forward(50)
alex.left(90)
alex.forward(50)
alex.left(90)


# REMOVED FOR GOOGLE COLAB:
# wn.exitonclick()
#
# Standard Turtle uses exitonclick() to keep its separate graphics
# window open until the user clicks it.
# In Google Colab, the drawing remains visible in the notebook,
# so this command is not needed.
````

## Example 4 — Repeating Instructions with a `for` Loop

The square requires the same two actions four times:

```python
alex.forward(50)
alex.left(90)
```

Instead of writing those lines four times, the `for` loop repeats them for us.

```python
for i in [0, 1, 2, 3]:
```

The list contains four values, so the indented commands run **four times**.

### Try changing it
- Change `50` to resize the square.
- Change the turning angle.
- Add or remove values from `[0, 1, 2, 3]` and observe how many times the loop repeats.

**Prediction first:** before you run your modified code, sketch or describe what you think it will draw.

````python
# CHANGED: Use ColabTurtlePlus instead of Python's standard turtle library.
import ColabTurtlePlus.Turtle as turtle

# ADDED: Clear any previous Turtle drawing from the Colab output.
turtle.clearscreen()


wn = turtle.Screen()
alex = turtle.Turtle()


for i in [0, 1, 2, 3]:      # repeat four times
    alex.forward(50)
    alex.left(90)


# REMOVED FOR GOOGLE COLAB:
# wn.exitonclick()
#
# In standard Turtle, this keeps the graphics window open until
# the user clicks it. In Google Colab, the drawing remains visible
# in the notebook, so this line is not needed.
````

## Example 5 — Looping Through Colors

A `for` loop can repeat instructions using values from a list.

Here, `aColor` becomes each color in the list, one at a time:

```python
["yellow", "red", "purple", "blue"]
```

During each repetition, Alex changes color before drawing the next side.

### Try changing it
- Reorder the colors.
- Replace colors with your favorites.
- Change the side length.
- Add another color to the list. What happens to the drawing?
- Change the turn angle and look for a new pattern.

````python
# CHANGED: Use ColabTurtlePlus instead of Python's standard turtle library.
import ColabTurtlePlus.Turtle as turtle

# ADDED: Clear any previous Turtle drawing from the Colab output.
turtle.clearscreen()


wn = turtle.Screen()
alex = turtle.Turtle()


for aColor in ["yellow", "red", "purple", "blue"]:
    alex.color(aColor)
    alex.forward(50)
    alex.left(90)


# REMOVED FOR GOOGLE COLAB:
# wn.exitonclick()
#
# In standard Turtle, this keeps the graphics window open until
# the user clicks it. In Google Colab, the drawing remains visible
# in the notebook, so this line is not needed.
````

## Example 6 — `range()`, Stamps, and Patterns

This example creates a growing spiral-like pattern.

The expression

```python
range(5, 60, 2)
```

generates values beginning at `5`, increasing by `2`, and stopping before `60`. Each value becomes the variable `size`.

Two other important commands are:

- `tess.up()` — lifts the pen so Tess can move without drawing a line.
- `tess.stamp()` — leaves a copy of the Turtle shape at its current position.

### Try changing it
Change **one value at a time** and observe the result:
- `range(5, 60, 2)` → change the start, stop, or step.
- `tess.right(24)` → try a different turning angle.
- `tess.color("blue")` → choose another color.
- Try removing `tess.up()` temporarily. How does the picture change?

**Challenge:** Can you create a pattern that looks like a flower, starburst, or spiral?

````python
# CHANGED: Use ColabTurtlePlus instead of Python's standard turtle library.
import ColabTurtlePlus.Turtle as turtle

# ADDED: Clear any previous Turtle drawing from the Colab output.
turtle.clearscreen()


wn = turtle.Screen()
wn.bgcolor("lightgreen")

tess = turtle.Turtle()
tess.color("blue")
tess.shape("turtle")


print(list(range(5, 60, 2)))

tess.up()                         # lift the pen so tess moves without drawing

for size in range(5, 60, 2):      # start with size = 5 and grow by 2
    tess.stamp()                  # leave an impression on the canvas
    tess.forward(size)            # move tess along
    tess.right(24)                # and turn her


# REMOVED FOR GOOGLE COLAB:
# wn.exitonclick()
#
# In standard Turtle, this keeps the graphics window open until
# the user clicks it. In Google Colab, the drawing remains visible
# in the notebook, so this line is not needed.
````
